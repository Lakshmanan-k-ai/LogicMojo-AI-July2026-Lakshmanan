```python
class Date:
    def __init__(self, year, month, day):
        print ("Date constructor called")
        self.year, self.month, self.day = year, month, day

    @classmethod
    def from_string(cls, date_str):
        y, m, d = map(int, date_str.split("-"))
        print (cls)
        return cls(y, m, d)

d = Date.from_string("2026-07-29")
print (f"{d}") 
```

```python
class Account:
    def __init__(self, balance=0):
        self.balance = balance
    
    def summary (self):
        return (f"Account: {self.balance}")

class SavingsAccount(Account):
    def __init__(self, balance=0, interest_rate=0):
        super().__init__(balance)
        self.interest_rate = interest_rate

    def summary (self):
        return (f"Savings Account: {self.balance} at {self.interest_rate}")

a = Account(100);
sa = SavingsAccount(200, 0.05)
accounts = [a, sa];
for account in accounts:
    print (account.summary())

    
```

```python
raw = ["10", "7", "x", "3"]
value = 0
skipped = 0
for n in raw:
    try:
        value += int(n)
    except ValueError:
        skipped += 1
print (f"Total: {value}")
print (f"skipped: {skipped}")
```

```python
class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit (self, amount):
        self.balance += amount

a = Account(100)
a.deposit(50)
print (f"Account A balance: {a.balance}")

b = Account(0)
b.deposit(20)
print (f"Account B balance: {b.balance}")
```
