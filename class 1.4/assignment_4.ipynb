```python
import numpy as np
numbers = np.array([2, 4, 6, 8, 10])
print (f"plus 10: {numbers+10}")
a= np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(f"product: {a * b}")
print(f"mean: {numbers.mean()}")
```

```python
data = np.array([3, 8, 1, 10, 6, 4])
mean = data.mean()
print (f"mean: {mean:.2f}")

mask = data > mean
filtered_data = data[mask]
print (f"above mean: {filtered_data}")
print (f"count: {mask.sum()}")
```

```python
qty = np.array([2, 1, 3])
price = np.array([10, 25, 5])
total_cost = np.dot(qty, price)
print(f"dot product: {total_cost}")
total_cost = np.sum(qty * price)
print(f"sum: {total_cost}")
sum =0
for i in range(len(qty)):
    sum+= qty[i] * price[i] 
print(f"loop: {sum}")
print(f"match:{sum == total_cost}")
```
