```python
fruits = ["apple", "banana", "cherry"]

#Using for loop to iterate through the list
for i in range(len(fruits)):
    print(fruits[i])

#Using while loop to iterate through the list
n=0
while n in range(len(fruits)):
    print(fruits[n])
    n+=1
    
```
